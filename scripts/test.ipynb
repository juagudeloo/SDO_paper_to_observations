```python

fits_matched = []
nf = 0

for file in obs_files:

    nf += 1
    print(file)
    for i in range(len(test_event["phenomenon"])):

        tmp_info = {
                "timestamp_start": test_event["timestamp_start"],
                "timestamp_end": test_event["timestamp_end"],
                "instrument": test_event["instrument"],
                "wavelength_angstrom": test_event["wavelength_angstrom"],
                "phenomenon": test_event["phenomenon"][i],
                "center_tx_arcsec": test_event["center_tx_arcsec"][i],
                "center_ty_arcsec": test_event["center_ty_arcsec"][i],
                "observation_filename": test_event["observation_filenames"][i],
                "limb_position": test_event["limb_position"][i],
                "fov_arcsec": test_event["fov_arcsec"][i]
            }

        conf_details = {
            "filename": "",
            "box": [],
            "map": "",
            "observation_filename": test_event["observation_filenames"][i],
            "confidence": 0,
            "fetch_case": ""
        }

        #print(f"Starting the matching of the event {i}/{len(test_event["phenomenon"])} detected on the timeframe {tmp_info["timestamp_start"]} to {tmp_info["timestamp_end"]}")

        # First case scenario
        extracted = False
        
        smap = ft_exs.load_map(file)
        info_a = ft_exs.apply_strategy_a(smap, tmp_info)

        if info_a[-1] == True:
            smap = info_a[0]
            ([y], [x]), (h, w), conf = ft_exs.find_image(smap, tmp_info, metadata_dir)

            conf_details["filename"] = file
            conf_details["box"] = [([y], [x]), (h, w)]
            conf_details["map"] = smap
            conf_details["confidence"] = conf
            conf_details["fetch_case"] = "a"                

            #print(f"Extracted {nf}/{len(obs_files)} fits with confidence {conf} over CV2 evaluation")
            extracted = True

        # Second case scenario
        if extracted == False:
            info_b = ft_exs.apply_strategy_b(smap, tmp_info)

            if info_b[-1] == False:
                print("No events found")
                
            elif info_b[-1] == True:
                ec = 0
                extracted =  True
                for smap in info_b[0]:
                    ([y], [x]), (h, w), conf = ft_exs.find_image(smap, tmp_info, metadata_dir)
                    
                    conf_details["filename"] = file
                    conf_details["box"] = [([y], [x]), (h, w)]
                    conf_details["map"] = smap
                    conf_details["confidence"] = conf
                    conf_details["fetch_case"] = "b"

                    ec += 1
                    #print(f"Extracted {nf}/{len(obs_files)} and detected event {ec}/{len(info_b[0])} fits with confidence {conf} over CV2 evaluation")
            else:
                print(info_b)

        #Third case scenario
        if extracted == False:
            fitcase = "c"

            info_c = ft_exs.apply_strategy_c(smap, tmp_info)

            if info_c[-1] is False:
                fitcase = "d"

            smap = info_c[0]
            ([y], [x]), (h, w), conf = ft_exs.find_image(smap, tmp_info, metadata_dir)

            conf_details["filename"] = file
            conf_details["box"] = [([y], [x]), (h, w)]
            conf_details["map"] = smap
            conf_details["confidence"] = conf
            conf_details["fetch_case"] = fitcase

            #print(f"Extracted {nf}/{len(obs_files)} fits with confidence {conf} over CV2 evaluation")

        fits_matched.append(conf_details)
```

```python
def keep_highest_confidence(conf_details_list):
    """
    Group dicts by 'observation_filename' and keep only the one
    with the highest 'confidence' value in each group.
    """
    best_by_filename = {}

    for conf in conf_details_list:
        fname = conf["observation_filename"]
        if fname not in best_by_filename or conf["confidence"] > best_by_filename[fname]["confidence"]:
            best_by_filename[fname] = conf

    return list(best_by_filename.values())
```

```python
best_fits = keep_highest_confidence(fits_matched)
```

```python
from astropy.coordinates import SkyCoord
def get_true_corners(coord1, coord2):
    """
    Given two SkyCoord corners (in Helioprojective or similar frame),
    return the true bottom-left and top-right corners regardless of
    how they were originally labeled.
    """
    tx_vals = [coord1.Tx, coord2.Tx]
    ty_vals = [coord1.Ty, coord2.Ty]

    tx_min, tx_max = min(tx_vals), max(tx_vals)
    ty_min, ty_max = min(ty_vals), max(ty_vals)

    bottom_left = SkyCoord(tx_min, ty_min, frame=coord1.frame)
    top_right = SkyCoord(tx_max, ty_max, frame=coord1.frame)

    return bottom_left, top_right
```

```python
import matplotlib.colors as colors
import matplotlib.pyplot as plt
from sunpy.coordinates import SphericalScreen
from astropy import units as u
import cv2
import os

save_dir = "C:\\Users\\julio\\OneDrive - Universidad Nacional de Colombia\\Documentos\\Maestria\\Nacional\\Proyecto Juanes\\test_fits\\Song-Paper\\fits"
imgs_dir = metadata_dir

for fit in best_fits:

    total_map = ft_exs.load_map(fit["filename"])
    if "hmi" not in str(total_map.instrument).lower():
        total_map.plot_settings['norm'] = colors.LogNorm(1, total_map.max())

    sel_map = fit["map"]
    if "hmi" not in str(total_map.instrument).lower():
        sel_map.plot_settings['norm'] = colors.LogNorm(1, sel_map.max())

    [y],[x] = fit["box"][0]
    h, w = fit["box"][1]
    with SphericalScreen(sel_map.observer_coordinate, only_off_disk=True):
        corner1 = sel_map.pixel_to_world(x*u.pix, (y+h)*u.pix)
        corner2 = sel_map.pixel_to_world((x+w)*u.pix, y*u.pix)

    bottom_left, top_right = get_true_corners(corner1, corner2)
    opencv_sector = sel_map.submap(bottom_left=bottom_left, top_right=top_right)

    bottom_left, top_right = get_true_corners(sel_map.bottom_left_coord, sel_map.top_right_coord)

    fig = plt.figure(figsize=(20, 6))
    ax1 = fig.add_subplot(1, 4, 1, projection=total_map)
    total_map.plot(axes=ax1)
    total_map.draw_quadrangle(
        bottom_left,
        top_right=top_right,
        axes=ax1,
        edgecolor='red',
        linewidth=2
    )

    bottom_left, top_right = get_true_corners(opencv_sector.bottom_left_coord, opencv_sector.top_right_coord)
    ax2 = fig.add_subplot(1, 4, 2, projection=sel_map)
    sel_map.plot(axes=ax2)
    sel_map.draw_quadrangle(
        bottom_left,
        top_right=top_right,
        axes=ax2,
        edgecolor='red',
        linewidth=2
    )
    ax2.set_title("Structure Query Detection")

    ax3 = fig.add_subplot(1, 4, 3, projection=opencv_sector)
    opencv_sector.plot(axes=ax3)
    ax3.set_title("Structure OpenCV")

    original_image = cv2.imread(os.path.join(imgs_dir, fit["observation_filename"]))
    RGB_img = cv2.cvtColor(original_image, cv2.COLOR_BGR2RGB)
    ax4 = fig.add_subplot(1, 4, 4)
    ax4.set_title("Original Image")
    ax4.imshow(RGB_img)

    print(f"Open CV2 confidence: {fit["confidence"]}")

    plt.tight_layout()
    plt.savefig(os.path.join(save_dir, f"{fit["observation_filename"].replace(".png","")}-best-fit-confidence-f{fit["confidence"]:.2f}.png"))
    plt.show()
    opencv_sector.save(os.path.join(save_dir, f"{fit["observation_filename"].replace(".png","")}-best-fit-confidence-f{fit["confidence"]:.2f}.fits"), overwrite=True)
    plt.close()
    print("---------------------------------------------------")
```
